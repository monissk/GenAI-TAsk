```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

```python
# Task 1: PromptTemplate (No f-string)
template_1 = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

print(template_1.invoke({"topic": "Artificial Intelligence"}))
```

```python
# TASK 2: Multi-Input Prompt System
template_2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# Test cases
print(template_2.invoke({"topic": "AI", "audience": "beginners", "tone": "friendly"}))
print(template_2.invoke({"topic": "Python", "audience": "kids", "tone": "fun"}))
print(template_2.invoke({"topic": "Deep Learning", "audience": "engineers", "tone": "technical"}))
```

```python
# TASK 3: Prompt Variations Engine
# Teaching
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

# Interview
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)

# Storytelling
story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic = "Machine Learning"

print(teaching_template.invoke({"topic": topic}))
print(interview_template.invoke({"topic": topic}))
print(story_template.invoke({"topic": topic}))
```

```python
# TASK 4: ChatPromptTemplate System
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("user", "Explain {topic}")
])

# Example
messages = chat_template.invoke({
    "role": "teacher",
    "topic": "Neural Networks"
})

print(messages)
```

```python
# TASK 5: Input Validation Layer
def validate_inputs(audience, tone):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        print("Invalid audience → defaulting to 'beginner'")
        audience = "beginner"

    if tone not in valid_tone:
        print("Invalid tone → defaulting to 'formal'")
        tone = "formal"

    return audience, tone


# Test
audience, tone = validate_inputs("kids", "cool")
print(audience, tone)
```

```python
# TASK 6: Prompt Generator App
def generate_prompt(topic, audience, tone, style):

    audience, tone = validate_inputs(audience, tone)

    if style == "teaching":
        template = PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} clearly step by step for {audience} in a {tone} tone"
        )

    elif style == "interview":
        template = PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Ask 3 interview questions about {topic} for {audience} in a {tone} tone"
        )

    elif style == "storytelling":
        template = PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} as a story for {audience} in a {tone} tone"
        )

    else:
        return "Invalid style"

    return template.invoke({
        "topic": topic,
        "audience": audience,
        "tone": tone
    })


# Test
print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

```python
# TASK 7: Template Reusability Test
reuse_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_inputs = [
    {"topic": "AI", "audience": "beginner", "tone": "fun"},
    {"topic": "Python", "audience": "expert", "tone": "formal"},
    {"topic": "ML", "audience": "intermediate", "tone": "casual"},
    {"topic": "Data Science", "audience": "beginner", "tone": "formal"},
    {"topic": "Deep Learning", "audience": "expert", "tone": "technical"},
]

for inp in test_inputs:
    print(reuse_template.invoke(inp))
```
